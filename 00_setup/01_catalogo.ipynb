```python
%sql
-- Intenção: criar o catálogo isolado do projeto CNES
CREATE CATALOG IF NOT EXISTS lakehouse_cnes
COMMENT 'PD Implementação de Data Lakehouse com Delta Lake - CNES';
```

```python
%sql
-- Intenção: criar os schemas da arquitetura Medallion
CREATE SCHEMA IF NOT EXISTS lakehouse_cnes.bronze;
CREATE SCHEMA IF NOT EXISTS lakehouse_cnes.silver;
CREATE SCHEMA IF NOT EXISTS lakehouse_cnes.gold;
```

```python
%sql
-- Intenção: criar o volume de aterrissagem dos arquivos brutos (fonte do Auto Loader)
CREATE VOLUME IF NOT EXISTS lakehouse_cnes.bronze.raw_files;

-- Conclusão esperada: schemas bronze, silver, gold (+ default e information_schema)
SHOW SCHEMAS IN lakehouse_cnes;
```
